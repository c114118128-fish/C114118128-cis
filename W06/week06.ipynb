```python
# ==========================================================
# Mission 8: 建立小考成績欄位（向量化 + 表格輸出）
# 版本：最終輸出 60 筆、不含統計摘要
# ==========================================================
import numpy as np
import pandas as pd
from IPython.display import display

TARGET_N = 60          # ← 需要的資料筆數（要改就改這裡）

# ---------- 0-1. 示範資料產生器 ----------
def make_demo_data(n, seed=2024):
    """依課程欄位產生 n 筆示範資料，回傳 (df, rng)"""
    _rng = np.random.default_rng(seed)
    _df = pd.DataFrame({
        "student_id":        [f"S{i+1:03d}" for i in range(n)],
        "study_hours":       np.round(_rng.uniform(0, 10, n), 1),   # 0~10 小時
        "sleep_hours":       np.round(_rng.uniform(4, 10, n), 1),   # 4~10 小時
        "screen_time_hours": np.round(_rng.uniform(0, 8,  n), 1),   # 0~8 小時
        "attendance_rate":   np.round(_rng.uniform(50, 100, n), 1), # 50~100 尺度
    })
    return _df, _rng

# ---------- 0-2. 前置檢查：df 筆數必須等於 TARGET_N ----------
if "df" not in globals():
    print(f"⚠️ 找不到 df → 自動產生 {TARGET_N} 筆示範資料")
    df, rng = make_demo_data(TARGET_N)
else:
    if len(df) > TARGET_N:
        print(f"ℹ️ 現有 df 有 {len(df)} 筆 → 保留前 {TARGET_N} 筆")
        df = df.head(TARGET_N).reset_index(drop=True)
    elif len(df) < TARGET_N:
        print(f"⚠️ 現有 df 只有 {len(df)} 筆，不足 {TARGET_N} 筆 → 重新產生")
        df, rng = make_demo_data(TARGET_N)
    else:
        print(f"✅ df 已是 {TARGET_N} 筆，直接沿用")

if "rng" not in globals():
    rng = np.random.default_rng(2024)   # 固定種子，結果可重現

# 確認必要欄位都齊全
need = ["student_id", "study_hours", "sleep_hours",
        "screen_time_hours", "attendance_rate"]
missing = [c for c in need if c not in df.columns]
if missing:
    raise KeyError(f"df 缺少欄位：{missing} → 請先執行前面建立資料的儲存格")

n = len(df)

# ---------- 1. 各項目對成績的貢獻（向量化）----------
base               = 50
study_contrib      =  3    * df["study_hours"]
sleep_contrib      =  2    * df["sleep_hours"]
screen_contrib     = -1.2  * df["screen_time_hours"]
attendance_contrib =  0.15 * df["attendance_rate"]

# 用獨立種子產生誤差，避免 rng 被前面儲存格用過後序列漂移
rng_err            = np.random.default_rng(2024)
random_error       = np.round(rng_err.normal(0, 5, n), 1)

# ---------- 2. 計算總分 ----------
df["quiz_score"] = (base + study_contrib + sleep_contrib
                    + screen_contrib + attendance_contrib + random_error)

# ---------- 3. 限縮 0~100、四捨五入成整數 ----------
df["quiz_score"] = np.clip(df["quiz_score"], 0, 100).round().astype(int)

# ---------- 4. 表格輸出：完整 60 筆 ----------
cols = ["student_id", "study_hours", "sleep_hours",
        "screen_time_hours", "attendance_rate", "quiz_score"]

print(f"資料筆數：{len(df)}\n")
display(df[cols].round(1))
```
